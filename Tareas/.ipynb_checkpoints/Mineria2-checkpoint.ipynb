```python
class def_matriz:
  def __init__(self, size):
    self.size = size

    self.matriz = [[2 for j in range(self.size)] for i in range(self.size)]

class producto:
  def __init__(self, size, A, B):
    self.A = A.matriz
    self.B = B.matriz
    self.size = size

    P = [[0 for j in range(self.size)] for i in range(self.size)]

    for i in range(self.size):
      for j in range(self.size):

          for n in range(self.size):
            P[i][j] += self.A[i][n]*self.B[n][j]
    '''
    for i in P:
      for p in i:
        print(p, end=" | ")
      print()
    print()
    '''
```

```python
import numpy as np

class def_matriznp:
  def __init__(self, size):
    self.size = size

    self.matriz = np.full((size, size), 2)

class producto_np:
  def __init__(self, size, A, B):
    self.A = A.matriz
    self.B = B.matriz
    self.size = size

    P = self.A @ self.B

    '''
    for i in P:
      for p in i:
        print(p, end=" | ")
      print()
    print()
    '''
```

```python
import time

n = [1, 5, 10, 50, 100, 500, 1000]
tiempo = []
for i in range(7):
  inic = time.time()

  A = def_matriz(n[i])
  B = def_matriz(n[i])
  result1 = producto(n[i], A, B)

  fin = time.time()
  tiempo.append(fin-inic)

for i in range(7):
  print(f"t({i+1})= {tiempo[i]}")
```

```python
tiempo_np = []
for i in range(7):
  inicnp = time.time()

  A = def_matriznp(n[i])
  B = def_matriznp(n[i])
  result1 = producto_np(n[i], A, B)

  finnp = time.time()
  tiempo_np.append(finnp-inicnp)

for i in range(7):
  print(f"t({i+1})= {tiempo_np[i]}")
```

```python
import matplotlib.pyplot as plt

plt.plot(n, tiempo, marker='o', label="Clases")
plt.plot(n, tiempo_np, marker='o',  label="Numpy")

plt.xscale('log')

plt.xlabel("n")
plt.ylabel("t")
plt.title("Comparación")

plt.legend()
plt.grid(True)
plt.show()

```
