# Limpieza de Datos
## Datos Nulos

Se eliminan las filas de la calle 19

```python
df1c = df1[df1['Acceso_Estacion'].notnull()]
df1c.reset_index()
```

```python
df1c.info()
```

Ahora hay que eliminar los registros de horas fuera del horario de operación.

Accederemos a las horas que se necesitan borrar desde el dataframe principal (df1) para borrar los del dataframe limpio (df1c)

```python
df1_clean = df1c.copy()

#Eliminar las horas de 00:00:00 a 03:45:00
for i in range(0, 9):
    df1_clean = df1_clean[df1_clean['Intervalo'] != df1.loc[i, "Intervalo"]]
```

```python
#Eliminar las horas de 23:15:00 a 23:45:00
for i in range(86, 89):
    df1_clean = df1_clean[df1_clean['Intervalo'] != df1.loc[i, "Intervalo"]]
```

```python
df1_clean = df1_clean[ (df1_clean['Intervalo'] != '02:30:00') & (df1_clean['Intervalo'] != '02:45:00') ]
```

```python
df1_clean['Intervalo'].unique()
```

Ahora solo se contienen los datos de las horas en las que opera transmilenio

Por último hay que eliminar las estaciones de Trasmicable

```python
df1_clean = df1_clean[ (df1_clean['Estacion'] != '(40002)Manitas') & 
                       (df1_clean['Estacion'] != '(40001)Juan Pablo II') &
                       (df1_clean['Estacion'] != '(40003)Mirador del Paraiso') ]
df1_clean.reset_index()
```

## Datos Atípicos
A pesar de haber borrado las horas fuera del horario, aun se observan múltiples valores atípicos.

Serán limpiados, teniendo en cuenta que esto puede borrar por completo el comportamiento de las horas pico en el servicio.

```python
plt.figure(figsize=(10, 6))
sns.boxplot(data=df1_clean, x="total_general", y="Linea")
plt.title("Salidas totales registradas")
```

```python
df1_outlier = df1_clean.copy()

def cleaning(df1_clean, df1_outlier, column):
    # Calcular cuartiles
    Q1 = df1_clean.iloc[:, column].quantile(0.25)
    Q3 = df1_clean.iloc[:, column].quantile(0.75)
    # Rango Intercuartil
    IQR = Q3 - Q1
    # Límites
    sup_limit = Q3 + 1.5 * IQR
    inf_limit = Q1 - 1.5 * IQR
    
    # Filtrar
    df1_outlier = df1_outlier[
        (df1_outlier.iloc[:, column] >= inf_limit) &
        (df1_outlier.iloc[:, column] <= sup_limit)
    ]
    return df1_outlier

for column in range(4, 35):
    df1_outlier = cleaning(df1_clean, df1_outlier, column)

plt.figure(figsize=(10, 6))
sns.boxplot(data=df1_outlier, x="total_general", y="Linea")
plt.title("Salidas totales registradas")
```

```python
df1_outlier.describe().T
```

```python
heatmap_data = df1_clean.pivot_table(
    index="Linea",       # Eje Y
    columns="Intervalo",    # Eje X
    values="total_general", # El color/valor
)

# 2. Configurar el tamaño del lienzo
plt.figure(figsize=(16, 8))

# 3. Crear el Heatmap con Seaborn
sns.heatmap(
    heatmap_data, 
    cmap="viridis", 
    cbar_kws={"label": "Salidas Totales registradas"}
)

# 4. Ajustar el tamaño y rotación de las letras para que se lean bien
plt.xticks(rotation=90, fontsize=10) 
plt.yticks(fontsize=10)

plt.tight_layout()
plt.show()
```
