# Carga de Datos

```python
import warnings
warnings.filterwarnings("ignore", category=FutureWarning)
```

```python
#Obtención de la data
from LHC import experiment_CMS
experiment_CMS(num_img=500, size=64, zip_name="CERN_dataset")
```

```python
import os
import zipfile
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from PIL import Image

def cargar_datos(zip_path="CERN_dataset.zip", carpeta_salida="CERN_datos"):
    #1. Proceso para descomprimir el archivo ZIP
    with zipfile.ZipFile(zip_path, 'r') as zip_ref:
        zip_ref.extractall(carpeta_salida)
    X = []
    y = []
    #2. Recorrer las carpetas con las imagenes
    for clase in os.listdir(carpeta_salida):
        clase_path = os.path.join(carpeta_salida, clase)
        if os.path.isdir(clase_path):
            for img_name in os.listdir(clase_path):
                img_path = os.path.join(clase_path, img_name)

                # 3. Procedimiento para cargar imagenes en escala de grises
                img = Image.open(img_path).convert("L")

                #Normalizar las matrices
                #X solo tiene puntos en negro (0) y en blanco (255)
                
                X.append(np.array(img)/255)
                y.append(clase)  # etiqueta = nombre de la carpeta
                
    X = np.array(X)
    return X, y
```

# Exploración Inicial de Datos

## Dimensiones de los Datos

```python
X, y = cargar_datos("CERN_dataset.zip")
print(f"Se cargaron {len(X)} imágenes.")
print(f"Clases encontradas:")
print(np.unique(y, return_counts=True))
print("Dimensiones de las imágenes:", X.shape)
```

Se poseen un total de 3000 imágenes de 64x64 píxeles con 500 en cada clase, dado que las instancias están equitativamente distribuidas en las clases la métrica Accuracy no presenta problemas.

```python
def mostrar_imagenes(X, y, n):
    indices = np.random.choice(len(X), n, replace=False)  # elegir imagen aleatoriamente
    plt.figure(figsize=(12, 4))

    for i, idx in enumerate(indices):
        plt.subplot(1, n, i+1)
        plt.imshow(X[idx], cmap="gray")
        plt.title(y[idx])
        plt.axis("off")
    plt.show()

mostrar_imagenes(X, y, 8)
```

## Matrices Normalizadas

Las instancias de X se normalizaron durante la carga de datos

```python
np.unique(X)
```

## Reducción de Dimensionalidad 

```python
from sklearn.decomposition import PCA
from sklearn.preprocessing import LabelEncoder

# Pasar las matrices a bidimensionales
X_bi = X.reshape(len(X), 64 * 64)

# Codificar las clases para que puedan ser leidas por sklearn
le = LabelEncoder()
y_num = le.fit_transform(y)
nombres_clases = le.classes_

# Reducir las dimensiones hasta conservar el 95% de la varianza de los datos (no perder información de su distribución)
pca = PCA(n_components = 0.95)
X_r = pca.fit_transform(X_bi)
```

## Partición de Datos

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X_r,
                                                    y_num,
                                                    test_size=0.30,
                                                    random_state=40,
                                                    stratify=y_num)
```

```python
print(np.unique(y_test, return_counts=True))
```

# Modelo SVM

## SVM

```python
from sklearn.svm import SVC
from sklearn.metrics import classification_report
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt


kernels = ['linear', 'rbf', 'poly']
#guardar el modelo de cada kernel
modelos = {} 
```

En este modelo se implementó GridSearchCV, aplicando validación cruzada para la búsqueda de hiperparámetros apropiados, priorizando el accuracy. En este caso se hizo para los valores de C = 0.1, 1, 10, 100.

Se evaluó para los 3 kernel del modelo y se mostró su resultado. Dado que tardaba mucho tiempo en compilar simplemente se dejó el valor de C que presentaba un mejor resultado (C=0.1).

### Entrenamiento

```python
for kernel in kernels:
    model = SVC(kernel=kernel, C=0.1, probability=True, random_state=40) #probability necesario para la curva ROC
    model.fit(X_train, y_train)
    modelos[kernel] = model
    y_pred_train = model.predict(X_train)
    
    print("Kernel: ", kernel)
    print(classification_report(y_train, y_pred_train))

    # Calcular la matriz matemática
    cm = confusion_matrix(y_train, y_pred_train)
    
    # Renderizar la matriz visualmente
    disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=nombres_clases)
    disp.plot(cmap='Blues', values_format='d')
    plt.title(f"Matriz de Confusión - SVM - {kernel}")
    plt.show()
```

### Test

```python
for kernel in kernels:
    model = SVC(kernel=kernel, C=0.1, probability=True, random_state=40) #probability necesario para la curva ROC
    model.fit(X_train, y_train)
    modelos[kernel] = model
    y_pred = model.predict(X_test)
    
    print("Kernel: ", kernel)
    print(classification_report(y_test, y_pred))

    # Calcular la matriz matemática
    cm = confusion_matrix(y_test, y_pred)
    
    # Renderizar la matriz visualmente
    disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=nombres_clases)
    disp.plot(cmap='Blues', values_format='d')
    plt.title(f"Matriz de Confusión - SVM - {kernel}")
    plt.show()
```

## Curva ROC

Esta curva aplica para problemas de clasificación binaria, para implementarla en este caso hay que realizar una curva para cada clase.

El AUC nos daría información de la capacidad del modelo para discriminar a una partícula en específico de todas las demás.

```python
from sklearn.metrics import roc_curve, auc
from sklearn.preprocessing import label_binarize
import matplotlib.pyplot as plt

# Binarizar las clases
# 1 si pertenece a la clase, 0 si pertenece a cualquier otra
y_test_bin = label_binarize(y_test, classes=[0, 1, 2, 3, 4, 5])

# Asignarle un punta probabilistico a cada instancia para hacer la curva
y_score = modelos['linear'].predict_proba(X_test)

plt.figure(figsize=(10, 8))

# Calcular y graficar la curva ROC para cada partícula respecto a las demas
for i in range(6):
    fpr, tpr, _ = roc_curve(y_test_bin[:, i], y_score[:, i])
    roc_auc = auc(fpr, tpr)
    plt.plot(fpr, tpr, lw=2, label=f'{nombres_clases[i]} (AUC = {roc_auc:.2f})')

#linea para el caso de discriminación aleatoria
plt.plot([0, 1], [0, 1], color='gray', linestyle='--')

plt.xlabel('Tasa de Falsos Positivos')
plt.ylabel('Tasa de Verdaderos Positivos')
plt.title('Curva ROC Para Cada Clase (SVM)')
plt.legend(loc="lower right")
plt.show()
```

# Modelo de Regresión Logística

## Regresión Logística

```python
from sklearn.linear_model import LogisticRegression

model2_hiper = LogisticRegression(random_state=40)
```

### Busqueda de hiperparámetros

Aquí se muestra la implementación de GridSearchCV para probar hiperparámetros colocados de forma arbitraria, en este caso solo se varía C

```python
from sklearn.model_selection import GridSearchCV

#configurar grilla de hiperparámetros
param_grid = {
    'C': [0.01, 0.1, 5, 10, 100]      # Regularización
}

#GridSearchCV busca el mejor conjunto de hiperparámetros
grid_search_L = GridSearchCV(
    estimator=model2_hiper,
    param_grid=param_grid,
    cv=5,                  # Validación cruzada (5 particiones)
    scoring='accuracy', # Métrica objetivo
    n_jobs=-1
)
```

Aquí se escoge el conjunto de parámetros con mejor accuracy y se define como modelo

```python
#Entrenar explorando todas las combinaciones
grid_search_L.fit(X_train, y_train)
print(f"Mejores hiperparámetros: {grid_search_L.best_params_}")

#Escoger el mejor conjunto de hiperparámetros
model2 = grid_search_L.best_estimator_
```

## Matriz de Confusión

### Entrenamiento

```python
y_pred_train_L = model2.predict(X_train)

print(classification_report(y_train, y_pred_train_L))

# Calcular la matriz matemática
cm = confusion_matrix(y_train, y_pred_train_L)

# Renderizar la matriz visualmente
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=nombres_clases)
disp.plot(cmap='Blues', values_format='d')
plt.title("Matriz de Confusión - Regresión Logística")
plt.show()
```

### Test

```python
y_pred_L = model2.predict(X_test)

print(classification_report(y_test, y_pred_L))

# Calcular la matriz matemática
cm = confusion_matrix(y_test, y_pred_L)

# Renderizar la matriz visualmente
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=nombres_clases)
disp.plot(cmap='Blues', values_format='d')
plt.title("Matriz de Confusión - Regresión Logística")
plt.show()
```

# KNN

```python
from sklearn.neighbors import KNeighborsClassifier as KNN

model5_hiper = KNN()
```

### Busqueda de hiperparámetros

Para este caso se variaron 3 hiperparámetros:
- Número de vecinos más próximos a considerar
- Cual definición de distancia emplear (Manhattan o Euclideana)
- Cómo definir el peso o importancia de los vecinos próximos (todos con la misma importancia o inversamente proporcional a la distancia)

```python
#configurar grilla de hiperparámetros
param_grid = {
    'n_neighbors': [3, 4, 5, 6],                   # Número de vecinos cercanos
    'p': [1, 2],                                   # Métrica espacial (1: Distancia Manhattan, 2: Distancia Euclideana)
    'weights': ['uniform', 'distance']             # Importancia de los puntos uniforme entre todos o dependiente de la distancia
}

#GridSearchCV busca el mejor conjunto de hiperparámetros
grid_search_KNN = GridSearchCV(
    estimator=model5_hiper,
    param_grid=param_grid,
    cv=5,                  # Validación cruzada (5 particiones)
    scoring='accuracy', # Métrica objetivo
    n_jobs=-1
)
```

```python
#Entrenar explorando todas las combinaciones
grid_search_KNN.fit(X_train, y_train)
print(f"Mejores hiperparámetros: {grid_search_KNN.best_params_}")

#Escoger el mejor conjunto de hiperparámetros
model5 = grid_search_KNN.best_estimator_
```

## AUC

Con el objetivo de comparar la capacidad de los modelos de distinguir la trayectoria de una partícula en comparación con las demás a continuación se muestra el AUC calculado en la curva ROC para cada clase y modelo para el conjunto de datos de evaluación

```python
from sklearn.metrics import roc_auc_score

modelos_comp = {
    "SVM linear": modelos['linear'],
    "SVM rbf": modelos['rbf'],
    "SVM poly": modelos['poly'],
    "Reg. logística": model2,
    "Árbol": model3,
    "Naive Bayes": model4,
    "KNN": model5,
}

y_test_bin = label_binarize(y_test, classes=[0, 1, 2, 3, 4, 5])

# AUC de cada clase para cada modelo
aucs = {}
for nombre, m in modelos_comp.items():
    y_score = m.predict_proba(X_test)
    aucs[nombre] = [roc_auc_score(y_test_bin[:, i], y_score[:, i]) for i in range(6)]

df_auc = pd.DataFrame(aucs, index=nombres_clases)
df_auc.loc["Promedio (macro)"] = df_auc.mean()
print(df_auc.round(3))
```

```python
fig, ax = plt.subplots(figsize=(10, 5))
im = ax.imshow(df_auc.values, cmap="Blues", vmin=0.5, vmax=1.0, aspect="auto")

ax.set_xticks(range(df_auc.shape[1]))
ax.set_xticklabels(df_auc.columns, rotation=30, ha="right")
ax.set_yticks(range(df_auc.shape[0]))
ax.set_yticklabels(df_auc.index)

for i in range(df_auc.shape[0]):
    for j in range(df_auc.shape[1]):
        v = df_auc.values[i, j]
        ax.text(j, i, f"{v:.2f}", ha="center", va="center",
                color="white" if v > 0.8 else "black")

fig.colorbar(im, label="AUC")
ax.set_title("AUC por clase y modelo (test)")
plt.tight_layout()
plt.show()
```

# Discución y Conclusiones

```python
import random

def ej_clase(X, y, n_ej):
    for i in range(6):
        clase = np.where(y_num == i)[0]

        print(nombres_clases[i])
        for j in range(n_ej):
            idx = random.choice(clase)
            
            plt.subplot(1, n_ej, j+1)
            plt.imshow(X[idx], cmap="gray")
            plt.axis("off")
        plt.show()


ej_clase(X, y, n_ej = 8)
```

### Curva ROC (Capacidad de Distinción)

- La mayoría de los modelos distinguen apropiadamente las partículas quark y protones. Esto probablemente se debe a que poseen trayectorias notablemente distintas al resto de patículas, causando que la distribución de sus datos sea fácilmente diferenciable de las demas.
- Los modelos presentan el menor desempeño al momento de clasificar electrones y muones. Puede deberse a que:
    - Se puede observar que en algunas imágenes no se registró apropiadamente la trayectoria de las partículas, detectándose únicamente ruido.
    - La trayectoria de los muones está muy dispersa, lo que pudo favorecer a que fueran confundidas con ruido
    - Las trayectorias de las partículas tau y de los electrones son muy similares, dificultando su distinción
- El modelo KNN y SVM poly no diferencian las clases, clasifican las instancias de manera aleatoria al tener un AUC cercano o menor a 0.5

### Desempeño de los Kernel de SVM

- El kernel linear es el que presenta el mejor desempeño a la hora de clasificar las instancias de testeo.
    - Resultados perfectos para los quark y protones
    - Desempeño parcial para tau y neutrinos
    - Pero el modelo no clasifica correctamente muones y electrones, los confunde entre sí o con neutrinos y partículas tau
- El kernel polinómico es completamente inefectivo, solo logra clasificar correctamente algunas instancias de quarks, el resto los clasifica todos como electrones o muones.

### Aspectos a Mejorar

- Dada la similitud entre las trayectorias de muones, electrones y tau, se facilita la aparición de fallas a la hora de clasificarlas. Se requiere añadir algunas características extra que facilite la distinción.
- La inefectividad de algunos modelos no necesariamente sugiere su incapacidad de discriminación de las clases. Es probable que su bajo desempeño se deba a una elección inadecuada de hiperparámetros del modelo (como el parámetro de regularización C, o gamma), la cual se realizó de manera arbitraria. Puede que aumentar los valores de búsqueda en GridSearchCV pueda mejorar su desempeño.
- Basándose en los AUC y Accuracy, los modelos que mejor clasifican tienen un desempeño muy similar. Esta similitud evidencia que la baja efectividad no se debe al modelo en cuestión, si no a una falta de características en el dataset que concedan más información acerca de las trayectorias.
