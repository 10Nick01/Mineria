```python
class matriz:
  def __init__(self, filas, columnas):
    self.filas = filas
    self.columnas = columnas
    self.A = [[0 for j in range(self.columnas)] for i in range(self.filas)]

    print("Ingrese los elementos a(i,j)")

    #recorre todos los elementos fila por fila para darles valor
    for i in range(self.filas):
      for j in range(self.columnas):
        self.A[i][j] = float(input(f"a({i+1}, {j+1})= "))
      print()

    #muestra la matriz
    for i in self.A:
      for a in i:
        print(a, end=" | ")
      print()
    print()


  def traza(self):
    if self.filas != self.columnas:
      return "Error: La matriz no es cuadrada"
    else:
      self.tr = 0
      #sumar los elementos de la diagonal
      for i in range(self.filas):
        self.tr += self.A[i][i]
      return self.tr


  def escalar(self, numero):
    self.numero = numero
    M = [fila[:] for fila in self.A]

    for i in M:
      for m in i:
        m *= self.numero
        print(m, end=" | ")
      print()
    print()


  def traspuesta(self):
    aux_c = self.columnas
    aux_f = self.filas

    #intercambiar # de filas por # de columnas para matrices no cuadradas
    if self.columnas != self.filas:
      temp = aux_c
      aux_c = aux_f
      aux_f = temp

    self.A_t = [[0 for j in range(aux_c)] for i in range(aux_f)]

    for i in range(aux_f):
      for j in range(aux_c):
        self.A_t[i][j] = self.A[j][i]

    for i in self.A_t:
      for a in i:
        print(a, end=" | ")
      print()
    print()


  def determinante(self):
    if self.filas != self.columnas:
      return "Error: La matriz no es cuadrada"
    else:
      #copiar la matriz para que no se modifique
      M = [fila[:] for fila in self.A]
      n = self.filas
      self.det = 1

      for j in range(n):
        #buscar columna por columna, cual es el numero mayor al de la primera fila
        pivote = j
        for i in range(j+1, n):
            if abs(M[i][j]) > abs(M[pivote][j]):
                pivote = i

        #evitar divisiones entre 0
        #si uno de los pivotes es 0, al multiplicar al final el det da 0
        if abs(M[pivote][j]) < 1e-15:
            return 0

        #intercambiar filas para que el mayor esté arriba
        if pivote != j:
            M[j], M[pivote] = M[pivote], M[j]
            self.det *= -1

        #eliminación fila por fila
        for i in range(j+1, n):
            factor = M[i][j] / M[j][j]

            #realiar la operación para cada elemento de la fila
            for k in range(j, n):
                M[i][k] -= factor * M[j][k]

        #se multiplica por el elemento de la diagonal cuando los demás son 0
        self.det *= M[j][j]

      return self.det

  def inversa(self):
    if self.filas != self.columnas:
      print("Error: La matriz no es cuadrada")
    else:
      M = [fila[:] for fila in self.A]
      n = self.filas

      #definir matriz inversa
      A_i = [[0 for j in range(self.columnas)] for i in range(self.filas)]
      for i in range(n):
        A_i[i][i] = 1


      #matriz triangular superior
      for j in range(n):
        pivote = j
        for i in range(j+1, n):
            if abs(M[i][j]) > abs(M[pivote][j]):
                pivote = i

        if pivote != j:
          M[j], M[pivote] = M[pivote], M[j]
          A_i[j], A_i[pivote] = A_i[pivote], A_i[j]

        for i in range(j+1, n):
            factor = M[i][j] / M[j][j]

            for k in range(n):
                M[i][k] -= factor * M[j][k]
                A_i[i][k] -= factor * A_i[j][k]

      #guardar la triangular superior
      self.A_t_s = [fila[:] for fila in M]

      #matriz diagonal
      for j in range(n-1, -1, -1):

        for i in range(j-1, -1, -1):
            factor = M[i][j] / M[j][j]

            for k in range(n):
                M[i][k] -= factor * M[j][k]
                A_i[i][k] -= factor * A_i[j][k]

      #matriz identidad
      for i in range(n):
        pivote = M[i][i]
        for k in range(n):
          M[i][k] /= pivote
          A_i[i][k] /= pivote

      for i in A_i:
        for a in i:
          print(f"{a:.3f}", end=" | ")
        print()
      print()

  #mostrar la matriz triangular superior
  def triangular_s(self):
    for i in self.A_t_s:
      for a in i:
        print(f"{a:.3f}", end=" | ")
      print()
    print()
```

```python
filas = int(input("Número de filas: "))
columnas = int(input("Número de columnas: "))

matriz1 = matriz(filas, columnas)

print("Traspuesta: ")
matriz1.traspuesta()

print("Multiplicación por escalar")
numero = float(input("Escalar: "))
matriz1.escalar(numero)

print(f"Traza: {matriz1.traza()}")

print(f"Determinante: {matriz1.determinante()}\n")

print("Matriz Inversa: ")
matriz1.inversa()

print("Matriz triangular superior: ")
matriz1.triangular_s()

```
